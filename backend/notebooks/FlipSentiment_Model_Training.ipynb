## 📌 Step 0 — Setup Dependencies & NLTK Resources

```python
import re
import os
import time
import json
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support, f1_score,
    classification_report, confusion_matrix
)

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Ensure NLTK resources are available automatically
for resource in ['stopwords', 'wordnet', 'omw-1.4']:
    try:
        nltk.download(resource, quiet=True)
    except Exception as e:
        print(f"NLTK download notice: {e}")

# Preprocessing helper
stop_words = set(stopwords.words('english')) - {'not', 'no', 'nor', 'neither', 'never', 'barely', 'hardly'}
lemmatizer = WordNetLemmatizer()

def preprocess_text(text: str) -> str:
    if pd.isna(text) or not str(text).strip():
        return ""
    text_str = str(text).lower()
    text_str = re.sub(r'https?://\S+|www\.\S+', '', text_str)
    text_str = re.sub(r'[\?]+', ' ', text_str)
    text_str = re.sub(r'[^a-zA-Z\s]', ' ', text_str)
    words = text_str.split()
    tokens = [lemmatizer.lemmatize(w) for w in words if w not in stop_words and len(w) > 1]
    return " ".join(tokens)

print("✅ All dependencies and preprocessing functions initialized successfully!")
```

---  
## 📊 Step 1 — Load and Inspect Data

* Load the cleaned review dataset.
* Inspect shape, missing values, sample records, and class distribution.
* Perform validation and deduplication on target review text.

```python
# Auto-detect dataset file location (local workspace or Kaggle /kaggle/input)
dataset_candidates = [
    'cleaned_flipkart_reviews.csv',
    '../cleaned_flipkart_reviews.csv',
    '/kaggle/input/flipkart-product-customer-reviews-dataset/cleaned_flipkart_reviews.csv',
    '/kaggle/input/flipkart-reviews/cleaned_flipkart_reviews.csv',
    'Data.csv'
]

data_file = None
for path in dataset_candidates:
    if os.path.exists(path):
        data_file = path
        break

if not data_file:
    raise FileNotFoundError("Dataset file not found! Please ensure 'cleaned_flipkart_reviews.csv' is available.")

print(f"📂 Loading dataset from: {data_file}")
df = pd.read_csv(data_file)

print(f"Dataset Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")

# Resolve Text and Sentiment columns
text_col = 'cleaned_review' if 'cleaned_review' in df.columns else ('full_review' if 'full_review' in df.columns else 'Review')
target_col = 'Sentiment' if 'Sentiment' in df.columns else 'sentiment'

df['review'] = df[text_col].astype(str)
df['sentiment'] = df[target_col].astype(str).str.lower().str.strip()

# Drop invalid or non-matching classes
df = df[df['sentiment'].isin(['positive', 'negative', 'neutral'])].copy()
df = df.dropna(subset=['review']).copy()
df = df.drop_duplicates(subset=['review', 'sentiment']).copy()

print(f"\nCleaned Dataset Shape: {df.shape}")
print("\nSentiment Class Breakdown:")
dist = df['sentiment'].value_counts()
pct = df['sentiment'].value_counts(normalize=True) * 100
print(pd.DataFrame({'Count': dist, 'Percentage (%)': pct.round(2)}))

# Display Visualizing Class Distribution
plt.figure(figsize=(7, 4))
sns.countplot(x=df['sentiment'], order=['positive', 'neutral', 'negative'], palette=['#2ecc71', '#f1c40f', '#e74c3c'])
plt.title("Flipkart Reviews Sentiment Class Distribution", fontsize=13, fontweight='bold')
plt.xlabel("Sentiment Category")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

---  
## ✂️ Step 2 — Train / Test Split

* Split data into 80% training and 20% testing.
* Use `stratify=y` to preserve exact class proportion in both train and test splits.
* Use `random_state=42` for strict reproducibility.

```python
X = df['review']
y = df['sentiment']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(f"Training set size: {len(X_train)} reviews")
print(f"Testing set size:  {len(X_test)} reviews")

print("\nStratified Split Verification:")
split_check = pd.DataFrame({
    'Train (%)': y_train.value_counts(normalize=True).round(4) * 100,
    'Test (%)': y_test.value_counts(normalize=True).round(4) * 100
})
print(split_check)
```

---  
## 🔠 Step 3 — TF-IDF Feature Extraction

* Convert text into numerical TF-IDF feature vectors.
* Use unigrams and bigrams (`ngram_range=(1,2)`), `max_features=15000`, `sublinear_tf=True`.
* **Critical ML Rule**: Fit TF-IDF vectorizer **ONLY on training data** to prevent data leakage.

```python
print("⚡ Fitting TF-IDF Vectorizer on X_train...")
start_t = time.time()

vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=15000,
    min_df=2,
    max_df=0.9,
    sublinear_tf=True
)

# Fit on training data ONLY
X_train_tfidf = vectorizer.fit_transform(X_train)
# Transform test data WITHOUT fitting
X_test_tfidf = vectorizer.transform(X_test)

elapsed = time.time() - start_t
print(f"✅ Vectorization completed in {elapsed:.2f} seconds!")
print(f"Train TF-IDF Matrix Shape: {X_train_tfidf.shape}")
print(f"Test TF-IDF Matrix Shape:  {X_test_tfidf.shape}")
print(f"Extracted Vocabulary Size:  {len(vectorizer.vocabulary_)} terms")
```

---  
## 🤖 Step 4 — Train Multiple Models with GridSearchCV

Train and compare 4 distinct classification models:
1. **Multinomial Naive Bayes** (`MultinomialNB`)
2. **Logistic Regression** (`LogisticRegression`)
3. **Support Vector Classifier** (`CalibratedClassifierCV(LinearSVC)`)
4. **Random Forest Classifier** (`RandomForestClassifier`)

Hyperparameters are tuned using `GridSearchCV` with 3-fold cross-validation.

```python
model_configs = {
    'Multinomial Naive Bayes': {
        'model': MultinomialNB(),
        'params': {'alpha': [0.1, 0.5, 1.0]}
    },
    'Logistic Regression': {
        'model': LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'),
        'params': {'C': [0.5, 1.0, 5.0]}
    },
    'Support Vector Classifier': {
        'model': CalibratedClassifierCV(LinearSVC(C=1.0, class_weight='balanced', random_state=42, max_iter=2000)),
        'params': {}
    },
    'Random Forest': {
        'model': RandomForestClassifier(random_state=42, n_jobs=-1),
        'params': {'n_estimators': [100, 150], 'max_depth': [25, None]}
    }
}

trained_models = {}

for model_name, cfg in model_configs.items():
    print(f"\n⏳ Training & Tuning {model_name}...")
    start_t = time.time()
    if cfg['params']:
        grid = GridSearchCV(estimator=cfg['model'], param_grid=cfg['params'], cv=3, scoring='f1_weighted', n_jobs=-1)
        grid.fit(X_train_tfidf, y_train)
        best_clf = grid.best_estimator_
        print(f"   Best Parameters: {grid.best_params_}")
    else:
        best_clf = cfg['model']
        best_clf.fit(X_train_tfidf, y_train)
        print("   Used standard default hyperparameter settings.")
    
    elapsed = time.time() - start_t
    print(f"   Completed in {elapsed:.2f}s")
    trained_models[model_name] = best_clf
```

---  
## 📈 Step 5 & 6 — Model Evaluation & Best Model Selection

* Compute Accuracy, Precision, Recall, Weighted F1-Score, Macro F1-Score.
* Display comparative metrics table.
* Select best performing model based primarily on Weighted F1-Score.
* Plot model comparison charts and confusion matrix of the best model.

```python
results_list = []
eval_details = {}

for model_name, clf in trained_models.items():
    y_pred = clf.predict(X_test_tfidf)
    acc = accuracy_score(y_test, y_pred)
    prec, rec, f1_w, _ = precision_recall_fscore_support(y_test, y_pred, average='weighted', zero_division=0)
    _, _, f1_m, _ = precision_recall_fscore_support(y_test, y_pred, average='macro', zero_division=0)
    cm = confusion_matrix(y_test, y_pred, labels=['positive', 'neutral', 'negative'])
    
    eval_details[model_name] = {'cm': cm, 'pred': y_pred}
    results_list.append({
        'Model': model_name,
        'Accuracy': round(acc, 4),
        'Precision': round(prec, 4),
        'Recall': round(rec, 4),
        'F1-Score': round(f1_w, 4),
        'Macro F1': round(f1_m, 4)
    })

summary_df = pd.DataFrame(results_list).sort_values(by='F1-Score', ascending=False)

print("=" * 70)
print("🏆 MODEL EVALUATION COMPARISON TABLE")
print("=" * 70)
display(summary_df)

winning_row = summary_df.iloc[0]
best_model_name = winning_row['Model']
best_model = trained_models[best_model_name]

print(f"\n🌟 BEST MODEL SELECTED: {best_model_name}")
print(f"Primary Selection Criterion (Weighted F1-Score): {winning_row['F1-Score']}")

# Performance Bar Plot
plt.figure(figsize=(10, 4.5))
melted = summary_df.melt(id_vars=['Model'], value_vars=['Accuracy', 'F1-Score'], var_name='Metric', value_name='Score')
sns.barplot(data=melted, x='Model', y='Score', hue='Metric', palette=['#3498db', '#9b59b6'])
plt.title("Classifier Model Performance Comparison", fontsize=13, fontweight='bold')
plt.ylim(0.5, 1.0)
plt.ylabel("Score")
plt.tight_layout()
plt.show()

# Confusion Matrix Plot for Best Model
plt.figure(figsize=(6, 5))
sns.heatmap(eval_details[best_model_name]['cm'], annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Positive', 'Neutral', 'Negative'], yticklabels=['Positive', 'Neutral', 'Negative'])
plt.title(f"Confusion Matrix — Best Model ({best_model_name})", fontsize=12, fontweight='bold')
plt.xlabel("Predicted Sentiment")
plt.ylabel("Actual Sentiment")
plt.tight_layout()
plt.show()
```

## 🧪 Step 7 — Test Inference on New / Real-Time Flipkart Reviews

Define a standalone prediction function `predict_sentiment(review_text)` matching training preprocessing and test on 5 Flipkart customer review samples.

```python
def predict_sentiment(review_text: str):
    """Accepts raw review text and returns predicted sentiment class with confidence scores."""
    cleaned = preprocess_text(review_text)
    if not cleaned.strip():
        cleaned = str(review_text).lower().strip()
    
    transformed = vectorizer.transform([cleaned])
    pred = best_model.predict(transformed)[0]
    
    confidence = None
    if hasattr(best_model, "predict_proba"):
        try:
            probas = best_model.predict_proba(transformed)[0]
            confidence = {cls: float(p) for cls, p in zip(best_model.classes_, probas)}
        except Exception:
            confidence = None
            
    return pred, confidence

# Test Samples
sample_reviews = [
    "The cooling is fantastic and it works smoothly in hot weather. Absolutely love this air cooler!",
    "Terrible experience! The product stopped working after 2 days and customer service was unhelpful.",
    "Average quality. It is okay for a small room, nothing extraordinary.",
    "Super fast delivery by Flipkart and the product build quality is top notch!",
    "Received defective item. Water leaking from bottom and fan is making loud screeching noise."
]

print("🧪 INFERENCE TEST RESULTS:")
print("=" * 75)
for i, rev in enumerate(sample_reviews, 1):
    label, conf = predict_sentiment(rev)
    print(f"Sample #{i}: \"{rev}\"")
    print(f"   --> Predicted Sentiment: {label.upper()}")
    if conf:
        conf_fmt = ", ".join([f"{k}: {v*100:.1f}%" for k, v in conf.items()])
        print(f"   --> Confidence Scores:  [{conf_fmt}]")
    print("-" * 75)
```

## 💾 Step 8 & 9 — Save Model Artifacts for Backend Integration

Serialize the winning trained model and fitted TF-IDF vectorizer using `joblib` so they can be loaded by Python backends (FastAPI / Flask / Streamlit).

```python
os.makedirs('models', exist_ok=True)

model_filename = 'models/best_sentiment_model.pkl'
vectorizer_filename = 'models/tfidf_vectorizer.pkl'
metadata_filename = 'models/model_metadata.json'

joblib.dump(best_model, model_filename)
joblib.dump(vectorizer, vectorizer_filename)

metadata_content = {
    'best_model_name': best_model_name,
    'accuracy': float(winning_row['Accuracy']),
    'precision': float(winning_row['Precision']),
    'recall': float(winning_row['Recall']),
    'f1_score': float(winning_row['F1-Score']),
    'vocabulary_size': len(vectorizer.vocabulary_),
    'saved_at': time.strftime("%Y-%m-%d %H:%M:%S")
}

with open(metadata_filename, 'w') as f:
    json.dump(metadata_content, f, indent=4)

print("🎉 MODEL ARTIFACTS SAVED SUCCESSFULLY!")
print(f" 📦 Model File:      {os.path.abspath(model_filename)}")
print(f" 📦 Vectorizer File: {os.path.abspath(vectorizer_filename)}")
print(f" 📄 Metadata File:   {os.path.abspath(metadata_filename)}")

print("\n🔄 INFERENCE FLOW READY FOR BACKEND INTEGRATION:")
print("  Raw Customer Review text")
print("  └──> Text Preprocessing (preprocess_text)")
print("  └──> Fitted TF-IDF Vectorizer (tfidf_vectorizer.pkl)")
print("  └──> Trained Sentiment Model (best_sentiment_model.pkl)")
print("  └──> Predicted Sentiment Class (Positive / Neutral / Negative)")
```
